```python
duplicates_count = df.duplicated().sum()
print("Number of duplicates ", duplicates_count)
```

```python

df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
print( df['order_date'].dtype)
```

```python

df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.strftime('%b')
df['day_name'] = df['order_date'].dt.day_name()

df[['order_date', 'year', 'month', 'day_name']].head(3)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import  plotly.graph_objects as go
import plotly.figure_factory as ff

```

```python
df_plot = df.groupby('order_date')['total_revenue'].sum().reset_index().sort_values('order_date')

fig = px.line(df_plot, x='order_date', y='total_revenue', title='Total Revenue Over Time')
fig.update_xaxes(title_text='Order Date')
fig.update_yaxes(title_text='Total Revenue ($)')
fig.show()
```

```python
df_plot = df.groupby('product_category')['total_revenue'].sum().reset_index().sort_values('total_revenue', ascending=True)

fig = px.bar(df_plot, x='total_revenue', y='product_category', orientation='h', title='Total Revenue by Product Category')
fig.update_xaxes(title_text='Total Revenue ($)')
fig.update_yaxes(title_text='Product Category')
fig.show()
```

```python
df_plot = df.groupby('customer_region')['profit'].sum().reset_index().sort_values('profit', ascending=False)

fig = px.bar(df_plot, x='customer_region', y='profit', title='Total Profit by Customer Region')
fig.update_xaxes(title_text='Customer Region')
fig.update_yaxes(title_text='Total Profit ($)')
fig.show()
```

```python
df_plot = df['payment_method'].value_counts().reset_index()
df_plot.columns = ['payment_method', 'count']

fig = px.pie(df_plot, names='payment_method', values='count', title='Payment Methods Distribution')
fig.show()
```

```python
df_plot = df.groupby('product_category')['rating'].mean().reset_index().sort_values('rating', ascending=False)

fig = px.bar(df_plot, x='product_category', y='rating', title='Average Customer Rating by Category')
fig.update_xaxes(title_text='Product Category')
fig.update_yaxes(title_text='Average Rating (Out of 5)')
fig.show()
```

```python
df_plot = df.groupby('month')['total_revenue'].sum().reset_index()

fig = px.bar(df_plot, x='month', y='total_revenue', title='Total Revenue by Month')
fig.update_xaxes(title_text='Month')
fig.update_yaxes(title_text='Total Revenue ($)')
fig.show()
```

```python
df_plot = df.groupby('discount_percent')['quantity_sold'].sum().reset_index()

fig = px.scatter(df_plot, x='discount_percent', y='quantity_sold', title='Impact of Discount Percent on Quantity Sold')
fig.update_xaxes(title_text='Discount Percent (%)')
fig.update_yaxes(title_text='Total Quantity Sold')
fig.show()
```

```python
df_plot = df.groupby(['year', 'customer_region'])['profit'].sum().reset_index()

fig = px.bar(df_plot, x='year', y='profit', color='customer_region', barmode='group', title='Yearly Profit by Customer Region')
fig.update_xaxes(title_text='Year')
fig.update_yaxes(title_text='Total Profit ($)')
fig.show()
```
